```python
SURNAME = "Magtalas"
SEED_NUM = len(SURNAME)

# --- Exercise 2: Signal Processing & Analysis ---
raw_signal = f"{SURNAME}_Signal_#2026!"
processed = "".join([c for c in raw_signal if c.isalnum()]).upper()

letters = sum(c.isalpha() for c in processed)
digits = sum(c.isdigit() for c in processed)

print("--- EXERCISE 2 RESULTS ---")
print(f"Raw Signal: {raw_signal}")
print(f"Processed Signal: {processed}")
print(f"Letters: {letters} | Digits: {digits}\n")
```
